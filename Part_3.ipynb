 Setup Gemini API

```python
!pip -q install -U google-genai

from google.colab import userdata
from google import genai

api_key = userdata.get("OPENAI_API_KEY")  # your AIzaSy... key
if not api_key:
    raise ValueError("Missing Colab Secret: OPENAI_API_KEY")

client = genai.Client(api_key=api_key)
print("Gemini client ready ✅")
```

Exercise 3 — BEFORE Summary

```python
TEXT_TO_SUMMARIZE = """
Customer: I was charged twice for my subscription this month.
Agent: Sorry about that—can you confirm the email on the account and the last 4 digits of the card?
Customer: Email is user@email.com. Last 4 is 1234.
Agent: Thanks. I see two charges. I’ll refund one and it should post in 3–5 business days.
Customer: Great—will I still have access?
Agent: Yes, access stays active. You’ll get a confirmation email shortly.
"""

before_prompt = f"""
Summarize the text below as a customer support ticket summary.

Text:
{TEXT_TO_SUMMARIZE}

Write a short paragraph summary.
"""

before = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=before_prompt
).text

print("=== BEFORE SUMMARY ===\n")
print(before)
```

Exercise 3 — Self-Critique + AFTER Summary

```python
critique_prompt = f"""
You previously wrote this summary:

[BEFORE SUMMARY]
{before}

Now do SELF-REFLECTION and IMPROVE it using the criteria below.

CRITERIA (must follow all):
1) Accuracy: only include facts present in the text; do NOT invent details.
2) Clarity: easy to read; no vague wording.
3) Audience: written for a support team internal ticket (professional tone).
4) Length: 60–90 words total.
5) Format: output EXACTLY in this format:

Issues:
- (1–2 bullets)

Key Details:
- (2–4 bullets)

Next Actions:
- (1–3 bullets)

Then provide:
A) "CRITIQUE:" (3–6 bullets) explaining what the BEFORE summary missed or did poorly against the criteria.
B) "AFTER SUMMARY:" using the required format and meeting all criteria.
"""

after_response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=critique_prompt
).text

print("=== SELF-CRITIQUE + AFTER SUMMARY ===\n")
print(after_response)
```
