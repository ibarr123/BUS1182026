```python
!pip -q install -U google-genai

import os
from google.colab import userdata
from google import genai
from google.genai import types

# You stored your Google key under this name earlier:
api_key = userdata.get("OPENAI_API_KEY")
if not api_key:
    raise ValueError("Missing Colab Secret: OPENAI_API_KEY (put your AIzaSy... key there)")

# Create client (Gemini Developer API)
client = genai.Client(api_key=api_key)
```

```python
SYSTEM_PROMPT = """You are a professional customer support AI.

Rules:
- Be polite, calm, and efficient.
- Give step-by-step solutions.
- Ask clarifying questions only if necessary.
- Never ask for passwords, SSNs, or full card numbers.
- If you cannot solve, explain the best next step and what non-sensitive info to collect.
"""

# Create a chat session (this automatically keeps conversation memory)
chat = client.chats.create(
    model="gemini-2.5-flash",
    # Many installs support config here; if yours errors, remove config=... and it will still work.
    config=types.GenerateContentConfig(
        system_instruction=SYSTEM_PROMPT,
        temperature=0.3,
        max_output_tokens=500,
    ),
)

print("Customer Support AI (type 'exit' to stop)\n")

while True:
    msg = input("You: ").strip()
    if msg.lower() in {"exit", "quit"}:
        print("Done.")
        break

    resp = chat.send_message(msg)
    print("\nAgent:", resp.text, "\n")
```
